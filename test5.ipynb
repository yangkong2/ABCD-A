```python
a = []

for i in range(3):
    line = [] #안쪽 리스트로 사용할 빈 리스트 생성
    for j in range(2):
        line.append(j+i) #안쪽 리스트에 0 추가
    a.append(line) #전체 리스트에 안쪽 리스트를 추가

print(a)
```
