```python
print('화면에 직접 출력')
print('ab\'c')
print("ab'c")
print("doesn't")
print('does')
print('doesn\'t')
print("'string'")
print("\"string\"")
print('"string"')
```

```python
s1 = '화면에 직접 출력'
s2 = 'ab\’c'
s3 = 'does'
print(s1)
print(s2)
print(s3)
print(s1[0]) #첫 글
print(s2[1]) #두 번째 글자
print(s3[1:3])
print(s3[-1]) #마지막 글자
print(s1[-2]) #뒤에서 두 번째 글자
```

```python
a = 2
b = 3.1456
c = a+b
print(a)
print(b)
print(c)
print (round(c, 2)) #c를 소수점 두자리까지 반올림 후 표시
print("%.4f"%(c)) #c를 소수점 두자리까지 표시
print(a+b) #c로 치환
print(a,b,a+b,c)
d = 1e2 #지수 형태
e = 1e-2 #지수 형태
print(d)
print(e)
```

```python
item1 = '사과'
price1 = 1000
item2 = '바나나'
price2 = 500
print(item1, price1)
print(item2, price2)
```

```python
str1 = '{0}는 {1}원 입니다.'
print(str1.format(item1, price1))
print(str1.format(item2, price2))
```

```python
print(item1, price1, sep='*', end='/')
print(item2, price2)
```

```python
str2 = '%s는 %d원입니다.'
print(str2%(item1, price1))
print(str2%(item2, price2))
```
